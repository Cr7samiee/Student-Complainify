## PART 3: LOAD & SPLIT DATA (80/20)

The training pipeline merges the original dataset with any new live submissions,
shuffles everything, then splits 80% for training / 20% for testing.

```python
# Load the real dataset, show category balance, then do the same 80/20
# shuffle-split that retrain.py performs.

import csv, os, random
from collections import Counter

TRAIN_PATH = r'E:/Project-VI/workspace/ComplaintMgmtSystem/data/train_dataset.csv'
rows = list(csv.DictReader(open(TRAIN_PATH, encoding='utf-8')))

print('Total rows      :', len(rows))
print('Sample row      :', dict(list(rows[0].items())))

counts = Counter(r['category'] for r in rows)
print('\nCategory distribution:')
for cat, n in counts.most_common():
    print('  {:<22} {:>4}  ({:.1f}%)'.format(cat, n, n / len(rows) * 100))

random.shuffle(rows)                 # shuffle for a fair split
split = int(len(rows) * 0.8)
train_rows, test_rows = rows[:split], rows[split:]
print('\n80/20 split:')
print('  train rows    :', len(train_rows))
print('  test  rows    :', len(test_rows))
```
